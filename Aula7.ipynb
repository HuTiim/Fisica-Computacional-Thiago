```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10,10,50)
y = np.linspace(-12,12,100)
X, Y = np.meshgrid(x,y)
x0 , y0 = 3,1
x1 , y1 = -3,-1 
f = np.exp(-(X-x0)**2)*np.exp(-(Y-y0)**2) -2*np.exp(-(X-x1)**2)*np.exp(-(Y-y1)**2)

```

```python
plt.contour(x,y,f,levels = 15)
plt.gca().set_aspect(1)
plt.show()
```

```python
plt.contour(x,y,f,levels = [0.01,0.5,0.8,1.5,1.8])
plt.gca().set_aspect(1)
plt.show()
```

```python
plt.imshow(f, origin = 'lower', extent = [x[-0],x[-1],y[0],y[-1]],
           aspect = 1, interpolation = 'bicubic', vmin = -2, vmax = 2,
           cmap = 'bwr')
plt.colorbar(fraction=0.046, pad=0.04)
plt.contour(x,y,f,levels = 5, colors = 'black')
plt.ylim(-5,5)
plt.show()
```

```python
plt.pcolormesh(x,y,f, vmin= -2, vmax= 2, cmap="bwr", rasterized = True)
plt.gca().set_aspect(1)
plt.colorbar()
plt.contour(x,y,f,levels = 5, colors = 'black')
plt.ylim(-5,5)
plt.show()
```

```python
x = np.linspace(-1, 1, 10)
v = np.linspace(-1,1,10)

X, V = np.meshgrid(x,v)

UX = V
UY = -X
```

```python

plt.quiver(x, v, UX, UY)
plt.gca().set_aspect(1)
plt.show()
```

```python
plt.streamplot(x, v, UX, UY)
plt.gca().set_aspect(1)
plt.show()
```
