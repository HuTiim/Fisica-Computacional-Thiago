```python

def pi_monte_carlo(n):
    
    '''
     Primeiro importamos os pacotes que serão usados para a construção do código adiante definimos as variáveis n e c sendo n = o número de pontos que teremos dentro
    do nosso quadrado e c a variável que iremos contabilizar quantos pontos estão dentro do círculo, montamos a função e dentro estabelecemos uma estrutura de repetição
    percorrendo todos os valores de n gerando os pontos nas coordenadas x e y de forma aleatória e verificando se os mesmos estão dentro do círculo de raio 1 que está dentro do 
    do quadrado logo é calculado a estimativa do valor de pi sendo 4 vezes a quantidade de pontos dentro do nosso círculo.
        
    '''
    
    c = 0 
    
    for i in range(n):
        
        x = rng.rand()
        y = rng.rand()
                
        if x*x + y*y <= 1:
            c += 1
            
    estimativa_pi = (4*c)/n
    return estimativa_pi


    
```

```python
%timeit pi_monte_carlo(n)
```

```python
def pi_monte_carlo(n , retorna_ponto = False):
    
    d = 0 
    f = 0
    dentro = []
    fora = []
    
    
    for i in range(n):
        
        x = rng.rand()
        y = rng.rand()
        
        if x*x + y*y <= 1:
            d += 1
            dentro.append([x,y])
            estimativa_pi = (4*d)/n
        else:
            f += 1
            fora.append([x,y])
              
    if retorna_ponto:
        return estimativa_pi, np.array(dentro), np.array(fora)
    else:
        return estimativa_pi
```
