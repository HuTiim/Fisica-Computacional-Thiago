```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import numpy.random as rng
import timeit
```

Questão 1

```python
n = 100
```

```python


def pi_monte_carlolento(n):
    
    
    c = 0 
    
    for i in range(n):
        
        x = rng.rand()
        y = rng.rand()
                
        if x*x + y*y <= 1:
            c += 1
            
    estimativa_pi = (4*c)/n
    return estimativa_pi
```

```python

valores_n = [10,100,500,1000]
tempo = []

for N in valores_n:
    tempoexec = timeit.timeit(lambda: pi_monte_carlolento(N), number=100)
    tempo.append(tempoexec)
```

```python
linearização = np.polyfit(valores_n,tempo,1)

plt.plot(valores_n,tempo)
plt.scatter(valores_n,tempo)
plt.plot(valores_n, np.polyval(linearização, valores_n), c = "red")
plt.show()
```

```python
n = 10000
```

```python
def pi_monte_carlo(n , retorna_ponto = False):
    
    d = 0 
    f = 0
    dentro = []
    fora = []
    
    
    for i in range(n):
        
        x = rng.rand()
        y = rng.rand()
        
        if x*x + y*y <= 1:
            d += 1
            dentro.append([x,y])
            estimativa_pi = (4*d)/n
        else:
            f += 1
            fora.append([x,y])
              
    if retorna_ponto:
        return estimativa_pi, np.array(dentro), np.array(fora)
    else:
        return estimativa_pi
```

```python
pi , dentro , fora = pi_monte_carlo(n, retorna_ponto= True)
```

```python
plt.scatter(dentro[:, 0] , dentro[:, 1], c = 'blue', s = 2)
plt.scatter(fora[:, 0], fora[:, 1], c = 'red', s = 2 )
plt.show()
```

```python
valores_n = [10,100,500,1000]
tempo = []

for N in valores_n:
    tempoexec = timeit.timeit(lambda: pi_monte_carlo(N), number=100)
    tempo.append(tempoexec)
```

```python

linearização = np.polyfit(valores_n,tempo,1)
plt.plot(valores_n, np.polyval(linearização, valores_n), c = "red")
plt.plot(valores_n,tempo)
plt.scatter(valores_n,tempo)
plt.show()
```
