```python
def f(p,l):

    x = [0]

    for passos in range (l):
        r = rng.rand()

        if r < p:
            x.append(x[-1] + 1)
        else: 
            x.append(x[-1] - 1)
    return x
```

```python
def g(p,l,n):

    z = []

    for loop in range (n):
        x = f(p,l)
        z.append(x)

    z = np.array(z).T
    return z
```
