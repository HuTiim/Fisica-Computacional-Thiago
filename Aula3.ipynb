```python
import numpy as np
import numpy.random as rng
import matplotlib.pyplot as plt
```

```python
def f(p,l):

    x = [0]

    for passos in range (l):
        r = rng.rand()

        if r < p:
            x.append(x[-1] + 1)
        else: 
            x.append(x[-1] - 1)
    return x
```

```python
def g(p,l,n):

    z = []

    for loop in range (n):
        x = f(p,l)
        z.append(x)

    z = np.array(z).T
    return z
```

```python
plt.plot(f(0.55,1000))
plt.show()
```

```python
plt.hist(g(0.50, 1000, 1000)[-1], density=True)
plt.hist(g(0.52, 1000, 500)[-1], density=True)

plt.show()
```
