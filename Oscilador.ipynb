```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def euler(fun,tempos,y0):
    
    y = [y0]

    dt = tempos[1] - tempos[0] 
    
    for t in tempos[:-1]:
        y.append(y[-1] + fun(t,y[-1]) * dt)

    return np.array(y)
```

```python
tempos = np.linspace(0,2*np.pi,100)
y0 = np.array([1,0])
w0 = 2*np.pi
c = 0.5
```

```python
aceleracao = lambda x , v: -w0**2*x-2*c*w0*v
fun = lambda t, y: np.array([y[1] , aceleracao(y[0],y[1])]) 
```

```python
aa = euler(fun , tempos , np.array([1,0]))
aa.shape
```

```python
plt.plot(tempos , aa )
plt.show()
```
