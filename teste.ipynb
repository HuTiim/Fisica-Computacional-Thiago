```python
import numpy as np
import matplotlib.pyplot as plt
import numpy.random as rng
```

```python
def pi_monte_carlo(n, retorna_ponto = False):

    pontodentro = 0
    pontofora = 0
    dentro = []
    fora = []

    for i in range (n):
        x = rng.rand()
        y = rng.rand()

        if x*x + y*y <=1:
            pontodentro += 1
            dentro.append([x,y])
            estimativa_pi = (4*pontodentro)/n            
        else:
            pontofora += 1
            fora.append([x,y])
    if retorna_ponto:
        return estimativa_pi , np.array(dentro) , np.array(fora)
    else:
        return estimativa_pi
```

```python
n = 100

pi, dentro, fora = pi_monte_carlo(n, retorna_ponto=True)
```
