```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def euler(fun,tempos,y0):
    
    y = [y0]
    dt = tempos[1] - tempos[0] 
    
    for t in tempos[:-1]:
        y.append(y[-1] + fun(t,y[-1]) * dt)

    return y
```

```python
tempos = np.linspace(0,5,100)
fun = lambda t,v: 10-2*v
```

```python
va = euler(fun , tempos , 10)
vb = euler(fun , tempos , 5)
vc = euler(fun , tempos , 0)
vd = euler(fun , tempos , -5)
ve = euler(fun , tempos , -10)

```

```python
plt.plot(tempos , va)
plt.plot(tempos , vb)
plt.plot(tempos , vc)
plt.plot(tempos , vd)
plt.plot(tempos , ve)
plt.show()
```
